```python
# Importing NumPy

import numpy as np

print(np.__version__)
```

```python
# NumPy Array

forces = np.array([100, 200, 300, 400])

print(forces)
```

```python
# Vector Operation

areas = np.array([10, 10, 10, 10])

stress = forces / areas

print(stress)
```

```python
# Engineering Example
# Beam Stress Distribution

force = np.array([1000, 2000, 3000, 4000])

area = 0.01

stress = force / area

print(stress)
```

```python
# Matrix Operation

A = np.array([
    [2,1],
    [1,3]
])

print(A)

# Matrix Addition

B = np.array([
    [1,2],
    [3,4]
])

print(A + B)

# Matrix Multiplication

print(A @ B)
```

```python
# Solving Linear Equations
# 2x + y = 5
# x + 3y = 6

A = np.array([
    [2,1],
    [1,3]
])

B = np.array([5,6])

X = np.linalg.solve(A,B)

print(X)
```

```python
# Engineering Mini Project

# Spring Calculator
# F = kx

k = 500

x = np.array([0.01,0.02,0.03,0.04])

F = k*x

print(F)
```

```python
# Numerical Methods Connection

x = np.linspace(0,3,100)

f = x**3 - x - 2

print(f)
```
