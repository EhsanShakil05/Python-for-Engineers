```python
# Importing Matplotlib

import matplotlib.pyplot as plt

print("Matplotlib installed")
```

```python
# Plotting

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y)

plt.title('Force vs Displacement')
plt.xlabel('Displacement(m)')
plt.ylabel('Force (N)')

plt.grid(True)

plt.show
```

```python
# Stress-Strain Curve

import matplotlib.pyplot as curve

stress = [0,0.001, 0.002, 0.003, 0.004]
strain = [0, 50, 100, 150, 200]

curve.plot(strain, stress)

curve.title('Stress-Strain Graph')
curve.xlabel("Strain")
curve.ylabel("Stress (MPa)")

curve.grid(True)

curve.show()
```

```python
# Using NumPy with Graphs

import numpy as np
import matplotlib.pyplot as graph

k = 500

x = np.linspace(0,0.05,100)

F = k*x

graph.plot(x,F)

graph.title("Hooke's Law")
graph.xlabel("Deflection (m)")
graph.ylabel("Force (N)")

graph.grid(True)

graph.show()
```

```python
# Plotting f(x) = x^3 - x - 2

import numpy as eq
import matplotlib.pyplot as root

x = eq.linspace(0,3,500)

f = x**3 - x - 2

root.plot(x,f)

root.axhline(0)

root.title("Root Finding Function")
root.xlabel("x")
root.ylabel("f(x)")

root.grid(True)

root.show()
```

```python
# Scatter Plot

import matplotlib.pyplot as plt

load = [100,200,300,400,500]
deflection = [0.2,0.4,0.58,0.79,1.01]

plt.scatter(load,deflection)

plt.title("Load vs Deflection")
plt.xlabel("Load (N)")
plt.ylabel("Deflection (mm)")

plt.grid(True)

plt.show()
```

```python
# Mini Project
# Pump Performance Curve

import matplotlib.pyplot as plt

flow = [10,20,30,40,50]
head = [50,47,42,34,25]

plt.plot(flow,head)

plt.title("Pump Performance Curve")
plt.xlabel("Flow Rate")
plt.ylabel("Head")

plt.grid(True)

plt.show()
```
